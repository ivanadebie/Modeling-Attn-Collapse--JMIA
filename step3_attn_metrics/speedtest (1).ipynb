```python
import math
import ast

import pandas as pd
from tqdm import tqdm

CPD_PATH = "prepared_dataset_cpd_with_attn_metrics_windows512.csv"

# Allow pandas to skip malformed rows instead of raising an error
df = pd.read_csv(
    CPD_PATH,
    engine="python",        # more tolerant parser
    on_bad_lines="skip"     # skip any broken line (likely just the last one)
)

print("Rows:", len(df))
print("Columns:", len(df.columns))
df.head(2)
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "lmsys/longchat-13b-16k"

print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("CUDA device:", torch.cuda.get_device_name(0))

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Let H100 use TF32 for extra speed
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True

tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)

# LongChat / LLaMA usually has no pad_token; use eos as pad
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.bfloat16 if device.type == "cuda" else torch.float32,
    device_map="auto" if device.type == "cuda" else None,
    attn_implementation="eager",   # avoids the SDPA warning
)

model.eval()
print("Model loaded on device:", device)
```

```python
# ----- CONFIGS FOR SPEED / MEMORY ------

# prefix = gold text OR distractor text
MAX_PREFIX_TOKENS = 128     # gold / distractor tokens
MAX_CHUNK_TOKENS  = 64      # model_answer_chunk tokens
BATCH_SIZE        = 32      # good starting point for H100. Increase if VRAM allows.

bos_id = tokenizer.bos_token_id
sep_id = tokenizer.eos_token_id  # use eos as [SEP]
pad_id = tokenizer.pad_token_id

print("BOS:", bos_id, "SEP:", sep_id, "PAD:", pad_id)

def get_distractor_text(raw):
    """Turn the 'distractors' field into a single string."""
    if pd.isna(raw):
        return ""
    txt = str(raw)
    try:
        val = ast.literal_eval(txt)
        if isinstance(val, list):
            return " ".join(str(x) for x in val)
        return str(val)
    except Exception:
        return txt
```

```python
from torch.nn.utils.rnn import pad_sequence

def make_example(row_idx, prefix_text, chunk_text, kind):
    """
    Create one attention example:
      [BOS] prefix [SEP] chunk
    kind: "gold" or "distr"
    """
    prefix_ids = tokenizer.encode(prefix_text, add_special_tokens=False)[:MAX_PREFIX_TOKENS]
    chunk_ids  = tokenizer.encode(chunk_text,  add_special_tokens=False)[:MAX_CHUNK_TOKENS]

    if len(chunk_ids) == 0:
        return None  # nothing to compute for this chunk

    input_ids = []

    if bos_id is not None:
        input_ids.append(bos_id)

    pref_s = len(input_ids)
    input_ids.extend(prefix_ids)
    pref_e = len(input_ids)

    input_ids.append(sep_id)

    ch_s = len(input_ids)
    input_ids.extend(chunk_ids)
    ch_e = len(input_ids)

    return {
        "row_idx": row_idx,
        "kind": kind,              # "gold" or "distr"
        "input_ids": input_ids,
        "pref_s": pref_s,
        "pref_e": pref_e,
        "ch_s": ch_s,
        "ch_e": ch_e,
    }

examples = []

# 🔸 If you want to debug on a subset first, uncomment this:
# df_iter = df.head(500)
df_iter = df  # full dataset

for idx, row in tqdm(df_iter.iterrows(), total=len(df_iter), desc="Preparing examples"):
    chunk_text = str(row.get("model_answer_chunk", "") or "").strip()
    if not chunk_text:
        continue

    # GOLD side
    gold_text = str(row.get("gold_text_chunk", "") or "").strip()
    if (not gold_text) and ("gold_text" in row.index):
        gold_text = str(row.get("gold_text", "") or "").strip()

    if gold_text:
        ex_gold = make_example(idx, gold_text, chunk_text, kind="gold")
        if ex_gold is not None:
            examples.append(ex_gold)

    # DISTRACTOR side
    distr_text = get_distractor_text(row.get("distractors", "")).strip()
    if distr_text:
        ex_distr = make_example(idx, distr_text, chunk_text, kind="distr")
        if ex_distr is not None:
            examples.append(ex_distr)

print("Total attention examples:", len(examples))
```

```python
def run_batch(ex_batch):
    # Prepare padded input_ids
    tensors = [torch.tensor(e["input_ids"], dtype=torch.long) for e in ex_batch]
    padded = pad_sequence(tensors, batch_first=True, padding_value=pad_id)
    attention_mask = (padded != pad_id).long()

    padded = padded.to(device)
    attention_mask = attention_mask.to(device)

    with torch.no_grad():
        out = model(
            padded,
            attention_mask=attention_mask,
            output_attentions=True,
        )

    # out.attentions: tuple(num_layers) of [B, H, Q, K]
    last_attn = out.attentions[-1]   # [B, H, Q, K]

    batch_results = []
    for b, ex in enumerate(ex_batch):
        attn = last_attn[b]  # [H, seq_len, seq_len]
        pref_s, pref_e = ex["pref_s"], ex["pref_e"]
        ch_s, ch_e     = ex["ch_s"],   ex["ch_e"]

        seq_len = attn.size(-1)

        # restrict to queries = chunk tokens
        chunk_attn = attn[:, ch_s:ch_e, :]  # [H, chunk_len, seq_len]

        # ---- Average attention entropy over heads + chunk tokens ----
        probs = chunk_attn  # already softmax'ed
        entropy = -(probs * (probs + 1e-12).log()).sum(dim=-1)  # [H, chunk_len]
        att_entropy_avg = entropy.mean().item()

        # ---- Mass / max to prefix region (gold or distractors) ----
        mask_pref = torch.zeros(seq_len, device=device)
        if pref_e > pref_s:
            mask_pref[pref_s:pref_e] = 1.0

        mass = (chunk_attn * mask_pref.view(1, 1, -1)).sum(dim=-1)  # [H, chunk_len]
        mass_avg = mass.mean().item()
        mass_max = mass.max().item() if pref_e > pref_s else 0.0

        batch_results.append({
            "row_idx": ex["row_idx"],
            "kind": ex["kind"],  # "gold" / "distr"
            "att_entropy_avg": att_entropy_avg,
            "mass_avg": mass_avg,
            "mass_max": mass_max,
        })

    return batch_results


results = []

for i in tqdm(range(0, len(examples), BATCH_SIZE), desc="Running model in batches"):
    batch = examples[i:i+BATCH_SIZE]
    batch_res = run_batch(batch)
    results.extend(batch_res)

len(results)
```

```python
import pandas as pd

results_df = pd.DataFrame(results)
print("results_df shape:", results_df.shape)

# How many unique rows got any metrics?
print("Unique row_idx in results:", results_df["row_idx"].nunique())

# Gold vs distractor coverage
gold_rows = results_df[results_df["kind"] == "gold"]["row_idx"].nunique()
distr_rows = results_df[results_df["kind"] == "distr"]["row_idx"].nunique()
both_rows = results_df.groupby("row_idx")["kind"].nunique()
both_rows = (both_rows == 2).sum()

print("Rows with gold metrics:      ", gold_rows)
print("Rows with distractor metrics:", distr_rows)
print("Rows with BOTH gold & distr:", both_rows)
```

```python
import numpy as np

# Make sure df has row_idx
df["row_idx"] = df.index

# (You already did this, but it's okay to rerun)
results_df = pd.DataFrame(results)
print("results_df shape:", results_df.shape)
print(results_df.head())

# --- GOLD metrics per row ---
gold_df = (
    results_df[results_df["kind"] == "gold"]
    .groupby("row_idx", as_index=True)
    .agg({
        "att_entropy_avg": "mean",
        "mass_avg": "mean"
    })
    .rename(columns={
        "att_entropy_avg": "att_avg_entropy",
        "mass_avg": "cross_attention_mass_to_gold"
    })
)

# --- DISTRACTOR metrics per row ---
distr_df = (
    results_df[results_df["kind"] == "distr"]
    .groupby("row_idx", as_index=True)
    .agg({
        "mass_max": "max"
    })
    .rename(columns={
        "mass_max": "distractor_attention_max"
    })
)

print("gold_df shape:", gold_df.shape)
print("distr_df shape:", distr_df.shape)

# --- Merge back into main CPD df ---

# 1) Drop old stub columns if they exist
cols_to_drop = [
    "att_avg_entropy",
    "cross_attention_mass_to_gold",
    "distractor_attention_max",
]
df_clean = df.drop(columns=cols_to_drop, errors="ignore")

# 2) Set index to row_idx for joining
df_merged = df_clean.set_index("row_idx")

# 3) Join new attention metrics
df_merged = df_merged.join(gold_df, how="left")
df_merged = df_merged.join(distr_df, how="left")

# 4) Reset index
df_merged = df_merged.reset_index(drop=True)

# Sanity check
print("Merged shape:", df_merged.shape)
print(
    df_merged[[
        "att_avg_entropy",
        "cross_attention_mass_to_gold",
        "distractor_attention_max"
    ]].notna().sum()
)

df_merged.head(3)
```
